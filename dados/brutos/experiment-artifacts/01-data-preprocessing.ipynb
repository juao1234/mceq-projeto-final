## Validate the responses per task 
#### Encode:
* 0: Wrong Answer
* 1: Correct Answer

### Error Report

```python
# ER1P Error Report Pattern 1
print ("ER1P Error Report Pattern 1")
finishedSurveyData['ER1P'] = finishedSurveyData['ER1P'].replace({'Your user does not have the necessary privileges for this resource.':  '0'})
finishedSurveyData['ER1P'] = finishedSurveyData['ER1P'].replace({'You used invalid authentication credentials to access the resource.':  '1'})
finishedSurveyData['ER1P'].value_counts().sort_index()
```

```python
# ER1N Error Report NON-Pattern 1
print ("ER1N Error Report NON-Pattern 1")
finishedSurveyData['ER1N'] = finishedSurveyData['ER1N'].replace({'Your user does not have the necessary privileges for this resource.':  '0'})
finishedSurveyData['ER1N'] = finishedSurveyData['ER1N'].replace({'The resource could not be found.':  '0'})
finishedSurveyData['ER1N'] = finishedSurveyData['ER1N'].replace({'You used invalid authentication credentials to access the resource.':  '1'})
finishedSurveyData['ER1N'].value_counts().sort_index()
```

```python
# ER2P Error Report Pattern 2
print ("ER2P Error Report Pattern 2")
finishedSurveyData['ER2P'] = finishedSurveyData['ER2P'].replace({'The \'birthday\' field contained invalid input.':  '1'})
finishedSurveyData['ER2P'].value_counts().sort_index()
```

```python
# ER2N Error Report NON-Pattern 2
print ("ER2N Error Report NON-Pattern 2")
finishedSurveyData['ER2N'] = finishedSurveyData['ER2N'].replace({'The \'birthday\' field was not filled out.':  '0'})
finishedSurveyData['ER2N'] = finishedSurveyData['ER2N'].replace({'The \'name\' field contained forbidden characters.':  '0'})
finishedSurveyData['ER2N'] = finishedSurveyData['ER2N'].replace({'The \'birthday\' field contained invalid input.':  '1'})
finishedSurveyData['ER2N'].value_counts().sort_index()
```

### Parameter Tree

```python
# PT1P Parameter Tree Pattern 1
print ("PT1P Parameter Tree Pattern 1")
finishedSurveyData['PT1P'] = finishedSurveyData['PT1P'].replace({8.0: '0'})
finishedSurveyData['PT1P'] = finishedSurveyData['PT1P'].replace({4.0: '1'})
finishedSurveyData['PT1P'].value_counts().sort_index()
```

```python
# PT1N Parameter Tree NON-Pattern 1
print ("PT1P Parameter NON-Tree Pattern 1")
finishedSurveyData['PT1N'] = finishedSurveyData['PT1N'].replace({8.0: '0'})
finishedSurveyData['PT1N'] = finishedSurveyData['PT1N'].replace({4.0: '1'})
finishedSurveyData['PT1N'].value_counts().sort_index()
```

```python
# PT2P Parameter Tree Pattern 2
print ("PT2P Parameter Tree Pattern 2")
finishedSurveyData['PT2P'] = finishedSurveyData['PT2P'].replace({22: '0'})
finishedSurveyData['PT2P'] = finishedSurveyData['PT2P'].replace({2: '1'})
finishedSurveyData['PT2P'].value_counts().sort_index()
```

```python
# PT2N Parameter Tree NON-Pattern 2
print ("PT2N Parameter Tree NON-Pattern 2")
finishedSurveyData['PT2N'] = finishedSurveyData['PT2N'].replace({22: '0'})
finishedSurveyData['PT2N'] = finishedSurveyData['PT2N'].replace({2: '1'})
finishedSurveyData['PT2N'].value_counts().sort_index()
```

### Wish List
**Manual encoding** has to be done aftewards for not most common answers!

```python
# WL1P Wish List Pattern 1
print ("WL1P Wish List Pattern 1")
finishedSurveyData['WL1P'] = finishedSurveyData['WL1P'].replace({'const student = GET https://my-university.org:8080/students/$20?fields=studentName,birthday,postalCode':  '0'})
finishedSurveyData['WL1P'] = finishedSurveyData['WL1P'].replace({'const student = GET https://my-university.org:8080/students/20?fields=studentName,birthday,postalCode':  '1'})
finishedSurveyData['WL1P'].value_counts().sort_index()
```

```python
# WL1N Wish List NON-Pattern 1
print ("WL1N Wish List NON-Pattern 1")
finishedSurveyData['WL1N'] = finishedSurveyData['WL1N'].replace({'?':  '0'})
finishedSurveyData['WL1N'] = finishedSurveyData['WL1N'].replace({'Dunno':  '0'})
finishedSurveyData['WL1N'] = finishedSurveyData['WL1N'].replace({'const response = GET https://my-university.org:8080/students/20\nconst student = {\n  studentName: response.studentName,\n  birthday: response.birthday,\n  postalCode: response.postalCode\n}':  '1'})
finishedSurveyData['WL1N'].value_counts().sort_index()
```

```python
# WL2P Wish List Pattern 2
print ("WL2P Wish List Pattern 2")
finishedSurveyData['WL2P'] = finishedSurveyData['WL2P'].replace({'?':  '0'})
finishedSurveyData['WL2P'] = finishedSurveyData['WL2P'].replace({'Dunno':  '0'})
finishedSurveyData['WL2P'] = finishedSurveyData['WL2P'].replace({'const entry = GET https://my-calendar.io:8080/entries/167?fields=title,attendees,status':  '1'})
finishedSurveyData['WL2P'].value_counts().sort_index()
```

```python
# WL2N Wish List NON-Pattern 2
print ("WL2N Wish List NON-Pattern 2")
finishedSurveyData['WL2N'] = finishedSurveyData['WL2N'].replace({'?':  '0'})
finishedSurveyData['WL2N'] = finishedSurveyData['WL2N'].replace({'Dunno':  '0'})
finishedSurveyData['WL2N'] = finishedSurveyData['WL2N'].replace({'const response = GET https://my-calendar.io:8080/entries/167\nconst entry = {\n  title: response.title,\n  attendees: response.attendees,\n  status: response.status\n}':  '1'})
finishedSurveyData['WL2N'].value_counts().sort_index()
```

### Metadata Element

```python
# ME1P Metadata Element Pattern 1
print ("ME1P Metadata Element Pattern 1")
finishedSurveyData['ME1P'] = finishedSurveyData['ME1P'].replace({'The videos are sorted in ascending order regarding the published date.':  '0'})
finishedSurveyData['ME1P'] = finishedSurveyData['ME1P'].replace({'The videos are sorted in descending order regarding the published date.':  '1'})
finishedSurveyData['ME1P'].value_counts().sort_index()
```

```python
# ME1N Metadata Element NON-Pattern 1
print ("ME1N Metadata Element NON-Pattern 1")
finishedSurveyData['ME1N'] = finishedSurveyData['ME1N'].replace({'The videos are sorted in ascending order regarding the published date.':  '0'})
finishedSurveyData['ME1N'] = finishedSurveyData['ME1N'].replace({'The videos are sorted in descending order regarding the number of views.':  '0'})
finishedSurveyData['ME1N'] = finishedSurveyData['ME1N'].replace({'The videos are sorted in descending order regarding the published date.':  '1'})
finishedSurveyData['ME1N'].value_counts().sort_index()
```

```python
# ME2P Metadata Element Pattern 2
print ("ME2P Metadata Element Pattern 2")
finishedSurveyData['ME2P'] = finishedSurveyData['ME2P'].replace({'The products are sorted in ascending order regarding the shipping costs.':  '1'})
finishedSurveyData['ME2P'].value_counts().sort_index()
```

```python
# ME2N Metadata Element NON-Pattern 2
print ("ME2N Metadata Element NON-Pattern 2")
finishedSurveyData['ME2N'] = finishedSurveyData['ME2N'].replace({'The products are sorted in descending order regarding the title.':  '0'})
finishedSurveyData['ME2N'] = finishedSurveyData['ME2N'].replace({'The products are sorted in ascending order regarding the shipping costs.':  '1'})
finishedSurveyData['ME2N'].value_counts().sort_index()
```

### API Gateway
**Manual encoding** has to be done aftewards for not most common answers!

```python
# AG1P API Gateway Pattern 1
print ("AG1P API Gateway Pattern 1")
finishedSurveyData['AG1P'] = finishedSurveyData['AG1P'].replace({'APIGateway':  '1'})
finishedSurveyData['AG1P'] = finishedSurveyData['AG1P'].replace({'API gateway ':  '1'})
finishedSurveyData['AG1P'] = finishedSurveyData['AG1P'].replace({'API Gateway':  '1'})
finishedSurveyData['AG1P'] = finishedSurveyData['AG1P'].replace({'API Gateway,Service 3':  '1'})
finishedSurveyData['AG1P'] = finishedSurveyData['AG1P'].replace({'API Gateway, Service 3':  '1'})
finishedSurveyData['AG1P'] = finishedSurveyData['AG1P'].replace({'Service 3, API Gateway':  '1'})
finishedSurveyData['AG1P'].value_counts().sort_index()
```

```python
# AG1N API Gateway NON-Pattern 1
print ("AG1N API Gateway NON-Pattern 1")
finishedSurveyData['AG1N'] = finishedSurveyData['AG1N'].replace({'all':  '0'})
finishedSurveyData['AG1N'] = finishedSurveyData['AG1N'].replace({'Dunno':  '0'})
finishedSurveyData['AG1N'] = finishedSurveyData['AG1N'].replace({'Client 1, 2 and 3':  '1'})
finishedSurveyData['AG1N'] = finishedSurveyData['AG1N'].replace({'client 1, client 2, client 3':  '1'})
finishedSurveyData['AG1N'] = finishedSurveyData['AG1N'].replace({'Client 1, Client 2, Client 3':  '1'})
finishedSurveyData['AG1N'] = finishedSurveyData['AG1N'].replace({'Client 1, Client 2. Client 3':  '1'})
finishedSurveyData['AG1N'] = finishedSurveyData['AG1N'].replace({'Client 1, Client 2, Client 3, Service 3':  '1'})
finishedSurveyData['AG1N'] = finishedSurveyData['AG1N'].replace({'Service 3, Client 1, Client 2, Client 3':  '1'})
finishedSurveyData['AG1N'].value_counts().sort_index()
```

```python
# AG2P API Gateway Pattern 2
print ("AG2P API Gateway Pattern 2")
finishedSurveyData['AG2P'] = finishedSurveyData['AG2P'].replace({'?':  '0'})
finishedSurveyData['AG2P'] = finishedSurveyData['AG2P'].replace({'Dunno':  '0'})
finishedSurveyData['AG2P'] = finishedSurveyData['AG2P'].replace({'API Gateway':  '1'})
finishedSurveyData['AG2P'] = finishedSurveyData['AG2P'].replace({'API Gateway, Main Service':  '1'})
finishedSurveyData['AG2P'] = finishedSurveyData['AG2P'].replace({'API Gateway, MainService':  '1'})
finishedSurveyData['AG2P'] = finishedSurveyData['AG2P'].replace({'Main Service, API Gateway':  '1'})
finishedSurveyData['AG2P'] = finishedSurveyData['AG2P'].replace({'MainService, API Gateway':  '1'})
finishedSurveyData['AG2P'] = finishedSurveyData['AG2P'].replace({'MainService, APIGateway':  '1'})
finishedSurveyData['AG2P'].value_counts().sort_index()
```

```python
# AG2N API Gateway NON-Pattern 2
print ("AG2N API Gateway NON-Pattern 2")
finishedSurveyData['AG2N'] = finishedSurveyData['AG2N'].replace({'Dunno':  '0'})
finishedSurveyData['AG2N'] = finishedSurveyData['AG2N'].replace({'All orange clients.\nMaybe blue clients also, depends on the usage':  '0'})
finishedSurveyData['AG2N'] = finishedSurveyData['AG2N'].replace({'Web App 1, Mobile App1, Mobile App2':  '1'})
finishedSurveyData['AG2N'] = finishedSurveyData['AG2N'].replace({'Web App 1, Mobile App 1, Mobile App 2':  '1'})
finishedSurveyData['AG2N'].value_counts().sort_index()
```

### Request Bundle
**Manual encoding** has to be done aftewards for not most common answers!

```python
# RB1P API Gateway Pattern 1
print ("RB1P API Gateway Pattern 1")
finishedSurveyData['RB1P'] = finishedSurveyData['RB1P'].replace({'?':  '0'})
finishedSurveyData['RB1P'] = finishedSurveyData['RB1P'].replace({'Dunno':  '0'})
finishedSurveyData['RB1P'] = finishedSurveyData['RB1P'].replace({'GET https://my-retail-store.com/customers/$5,$9,$13,$24':  '0'})
finishedSurveyData['RB1P'] = finishedSurveyData['RB1P'].replace({'GET https://my-retail-store.com/customers/$5,$9,$13,$24;':  '0'})
finishedSurveyData['RB1P'] = finishedSurveyData['RB1P'].replace({'GET https://my-retail-store.com/customers/5/9/13/24':  '0'})
finishedSurveyData['RB1P'] = finishedSurveyData['RB1P'].replace({'GET https://my-retail-store.com/customers/5,9,13,24':  '1'})
finishedSurveyData['RB1P'].value_counts().sort_index()
```

```python
# RB1N API Gateway NON-Pattern 1
print ("RB1N API Gateway NON-Pattern 1")
finishedSurveyData['RB1N'] = finishedSurveyData['RB1N'].replace({'?':  '0'})
finishedSurveyData['RB1N'] = finishedSurveyData['RB1N'].replace({'Dunno':  '0'})
finishedSurveyData['RB1N'] = finishedSurveyData['RB1N'].replace({'GET https://my-retail-store.com/customers/$id=5,9,13,24':  '0'})
finishedSurveyData['RB1N'] = finishedSurveyData['RB1N'].replace({'GET https://my-retail-store.com/customers/5\nGET https://my-retail-store.com/customers/9\nGET https://my-retail-store.com/customers/13\nGET https://my-retail-store.com/customers/24':  '1'})
finishedSurveyData['RB1N'].value_counts().sort_index()
```

```python
# RB2P API Gateway Pattern 2
print ("RB2P API Gateway Pattern 2")
finishedSurveyData['RB2P'] = finishedSurveyData['RB2P'].replace({'?':  '0'})
finishedSurveyData['RB2P'] = finishedSurveyData['RB2P'].replace({'Dunno':  '0'})
finishedSurveyData['RB2P'] = finishedSurveyData['RB2P'].replace({'DELETE https://my-grocery.com/products/$12,$16,$27,$35':  '0'})
finishedSurveyData['RB2P'] = finishedSurveyData['RB2P'].replace({'DELETE https://my-grocery.com/products/12,16,27,35':  '1'})
finishedSurveyData['RB2P'] = finishedSurveyData['RB2P'].replace({'DELETE https://my-grocery.com/products/12,16,27,35\n':  '1'})
finishedSurveyData['RB2P'].value_counts().sort_index()
```

```python
# RB2N API Gateway NON-Pattern 2
print ("RB2N API Gateway NON-Pattern 2")
finishedSurveyData['RB2N'] = finishedSurveyData['RB2N'].replace({'?':  '0'})
finishedSurveyData['RB2N'] = finishedSurveyData['RB2N'].replace({'Dunno':  '0'})
finishedSurveyData['RB2N'] = finishedSurveyData['RB2N'].replace({'DELETE https://my-grocery.com/products/$12\nDELETE https://my-grocery.com/products/$16\nDELETE https://my-grocery.com/products/$27\nDELETE https://my-grocery.com/products/$35':  '0'})
finishedSurveyData['RB2N'] = finishedSurveyData['RB2N'].replace({'DELETE https://my-grocery.com/products/12\nDELETE https://my-grocery.com/products/16\nDELETE https://my-grocery.com/products/27\nDELETE https://my-grocery.com/products/35':  '1'})
finishedSurveyData['RB2N'].value_counts().sort_index()
```
